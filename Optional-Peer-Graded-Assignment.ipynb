## Task 3: Use the appropriate tests to answer the questions provided.



For each of the following questions;

- Is there a significant difference in median value of houses bounded by the Charles river or not? (T-test for independent samples)

- Is there a difference in Median values of houses (MEDV) for each proportion of owner occupied units built prior to 1940 (AGE)? (ANOVA)

- Can we conclude that there is no relationship between Nitric oxide concentrations and proportion of non-retail business acres per town? (Pearson Correlation)

- What is the impact of an additional weighted distance  to the five Boston employment centres on the median value of owner occupied homes? (Regression analysis)

Be sure to:

1. State your hypothesis.

2. Use α = 0.05

3. Perform the test Statistics.

4. State the conclusion from the test.

```python
# Is there a significant difference in median value of houses bounded by the Charles river or not? (T-test for independent samples)
chas_yes = df[df['CHAS'] == 1]['MEDV']
chas_no = df[df['CHAS'] == 0]['MEDV']
t_stat, p_value = scipy.stats.ttest_ind(chas_yes, chas_no)
print(f"T-statistic: {t_stat}, P-value: {p_value}")
if p_value < 0.05:
    print("There is a significant difference in median value of houses bounded by the Charles river or not.")
else:
    print("There is no significant difference in median value of houses bounded by the Charles river or not.")
    
```

```python
# Is there a difference in Median values of houses (MEDV) for each proportion of owner occupied units built prior to 1940 (AGE)? (ANOVA)
model = ols('MEDV ~ C(AGE_GROUP)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

```python
# the anova test results show whether there is a statistically significant difference in median values of houses across different age groups.

```

```python
# Can we conclude that there is no relationship between Nitric oxide concentrations and proportion of non-retail business acres per town? (Pearson Correlation)
correlation, p_value = scipy.stats.pearsonr(df['NOX'], df['INDUS'])
print(f"Pearson Correlation: {correlation}, P-value: {p_value}")   
```

```python
# p value less than 0.05 indicates that we can reject the null hypothesis and conclude that there is a significant relationship between Nitric oxide concentrations and proportion of non-retail business acres per town.   
```

```python
#  What is the impact of an additional weighted distance  to the five Boston employment centres on the median value of owner occupied homes? (Regression analysis)
import statsmodels.api as sm
X = df['DIS']
y = df['MEDV'] 
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
predictions = model.predict(X)
print(model.summary())
```
